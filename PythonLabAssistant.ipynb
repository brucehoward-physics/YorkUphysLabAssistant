# Now the code is setup, to use the text-to-speech helper, do the following:

First you must use the appropriate mechanism to create a controller instance for the hardware. See the Capacitors lab manual for details, as an example

```
scale_ops = Scale.ScoutSTX()
```

Then, this class can be controlled by the LabPartner class

```
scale = partner.MyLabMate(_class=scale_ops)
```

To set up the object, use `SETUP()`

```
scale.SETUP()
```

And to read the scale use `GET()`

```
scale.GET()
```

To close the connection use `CLOSE()`

```
scale.CLOSE()
```

Some parts also have a `SET()` functionality as well, e.g. to set a voltage.

## IMPORTANT!! The text to speech software sometimes runs into trouble if multiple things are attached at once. IF you need to use more than one device as text-to-speech in a lab session, please delete any currently active text-to-speech applications. The hardware itself should remain active to set back up. For example:

```
del scale

psu_ops = PSU.GPD3303D()
hv_ops = HV.HV_control(psu_ops)
actuator_ops = ACT.Actuator(DAQ_name, psu_ops)

actuator = partner.MyLabMate( _class=actuator_ops )
actuator.SETUP()
actuator.SET(0)
actuator.GET()
actuator.CLOSE()
del actuator

scale = partner.MyLabMate( _class=scale_ops )
scale.GET()
scale.CLOSE()
```
